```python
import numpy as np
```

```python
embedding = {}
```

```python
with open('glove.6B.100d.txt', encoding='utf-8') as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.array(values[1:], dtype = 'float32' )
        embedding[word] = vector
        
```

```python
len(embedding)
```

```python
embedding['king'][:10]
```

```python
from sklearn.mtrics.pairwise import cosine_similarity
```

```python
def similar(t1,t2):
    return cosine_similarity(
        embedding[t1].reshape(1,-1),
        embedding[t1].reshape(1,-1)
    )[0][0]
```

```python
similar('king', 'queen')
```

```python
similar('king','banana')
```
