```python
import re
```

```python
text = 'I love machine learning'
```

```python
pattern = 'love'
```

```python
print(re.search(pattern, text))
```

```python
data = 'machine learning'
```

```python
print(re.search('^machine', data)) # start with machine
```

```python
print(re.search(r"learning$", data)) # end with learning
```
