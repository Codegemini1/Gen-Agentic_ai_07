```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sentences = [
    'I love machine learning',
    'Machine learning is amazing',
    'I enjoy deep learning',
    'I like playing cricket',
    'Cricket is a great sport'
]
```

```python
# text --> vectors
vectorizer = TfidfVectorizer()
```

```python
vectors = vectorizer.fit_transform(sentences)
```

```python
vectors
```

```python
similarity_matrix = cosine_similarity(vectors)
```

```python
similarity_matrix
```

```python
query  = ['I love deep learing']
```

```python
query_vec = vectorizer.transform(query)
```

```python
score = cosine_similarity(query_vec,vectors)
```

```python
query[0]
```

```python
score
```

```python
score.argmax()
```

```python
import numpy as np
```

```python
movies = {
    'Avengers' :     [1,0,0,1],
    'Iron Man' :     [1,0,0,1],
    'The notebook' : [0,0,1,0],
    'Superbad' :     [0,1,0,0],
    'Titanic' :      [0,0,1,0]
}
```

```python
movies_names = list(movies.keys())
```

```python
movies_names
```

```python
movie_vectors = np.array(list(movies.values()))
```

```python
movie_vectors
```

```python
similarity_matrix = cosine_similarity(movie_vectors)
```

```python
similarity_matrix
```

```python
for i in range(len(movies_names)):
    for j in range(len(movies_names)):
        print(f'{movies_names[i]} vs {movies_names[j]} = {similarity_matrix[i][j]:.2f}')
```
