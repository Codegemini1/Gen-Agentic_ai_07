```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN
```

```python
X = np.array([
    [1,2,3],
    [2,3,4],
    [3,4,5],
    [4,5,6]
])
```

```python
y = np.array([4,5,6,7])
```

```python
X = X.reshape(X.shape[0], X.shape[1], 1)
```

```python
X.shape[0]
```

```python
X[0]
```

```python
X    # (samples, timesteps, features)
```

```python
Sample(4) --- timesteps (3) --- feature(1)
```

```python
model = Sequential()
```

```python
model.add(SimpleRNN(10, activation='tanh', input_shape=(3,1)))  # 3 timesteps ---> 1 feature
model.add(Dense(1))
```

```python
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
model.fit(X,y, epochs=4000, verbose=0)
```

```python
test = np.array([[5,6,7]])
```

```python
test = test.reshape(1,3,1)
```
