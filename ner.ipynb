```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Sundar Pichai announced a new Gemini AI model at Google headquartes in California on June 10, 2025.
"""
```

```python
doc = nlp(text)
```

```python
for ent in doc.ents:
  print(f'{ent.text:<30} {ent.label_}')
```

```python
from pygments import style
from spacy import displacy

displacy.render(
    doc,
    style='ent',
    jupyter=True
)
```
